### Understanding Lego Theme vs Lego Sets

Walk into a lego store and you will see their products organized by theme. Their theme include Star wars, Batman, and many more

<img src="https://i.imgur.com/aKcwkSx.png">

A lego **set** is a particular box of Lego product. Therefore a single theme typically has many different sets


```python
# Reading the sets dataset
sets = pd.read_csv('/content/sets.csv')
# Looking at the first 5 rows to get a sense of the data
sets.head()
```

 Answering some questions from the sets data

```python
# In which year were the first LEGO sets released and what were these sets called?
first_year_set_index = sets['year'].idxmin()
sets.iloc[first_year_set_index][['year', 'name']]
```

```python
# How many different products did the LEGO company sell in their first year of operation?
sets[sets['year'] == 1949]
```

```python
# What are the top 5 LEGO sets with the most number of parts? 
sets.sort_values('num_parts', ascending=False).head()
```

```python
# Now let's take a look at how many sets the LEGO company has published year-on-year. 
sets.groupby('year').count()['set_num']
```
