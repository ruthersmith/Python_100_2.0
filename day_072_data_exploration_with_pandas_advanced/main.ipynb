## Preliminary Data Exploration and Data Cleaning with Pandas

Now that we've got our data loaded into our dataframe,
we need to take a closer look at it to help us understand what it is we are working with. 
This is always the first step with any data science project. Let's see if we can answer the some question

```python
# How many rows does our dataframe have? 
df.shape # This gives us the number of rows and columns (rows, columns)
```

```python
# What are the labels for the columns? Do the columns have names?
df.columns
```

```python
# Are there any missing values in our dataframe? Does our dataframe contain any bad data?
# we're going to look for NaN (Not A Number) values in our dataframe. NAN values are blank cells 
# or cells that contain strings instead of numbers.
df.isna()
```

```python
# Also looking at the last few rows of the dataframe
df.tail()
```

```python
# Deleting the Last Row because it was junk data
clean_df = df.dropna()  
clean_df.tail()
```

```python
# Accessing a single column like 'starting median salary'
clean_df['Starting Median Salary']
```

```python
# Accessing multiple columns
clean_df[['Starting Median Salary', 'Mid-Career Median Salary']]
```

```python
# To find the highest starting salary we can simply chain the .max() method.
clean_df['Starting Median Salary'].max()
```
