## Sorting Values & Adding Columns

```python
# Trying to see the lowest risk majors
# A low-risk major is a degree where there is a small difference between the lowest and highest salaries. 
# In other words, if the difference between the 10th percentile and the 90th percentile earnings of your major is small, 
# then you can be more certain about your salary after you graduate.
spred_col = clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']

# adding a new column to the dataframe at particular index, you could also do clean_df['Spread'] = spred_col
clean_df.insert(5, 'Spread', spred_col) 
clean_df.head()
```

```python
# To see which degrees have the smallest spread, we can use the .sort_values() method.
low_risk = clean_df.sort_values('Spread')
low_risk[['Undergraduate Major', 'Spread']].head()
```

```python
# Majors with the Highest Potential
high_earn = clean_df.sort_values('Mid-Career 90th Percentile Salary', ascending=False)
high_earn[['Undergraduate Major', 'Mid-Career 90th Percentile Salary']].head()
```

```python
# Majors with the Greatest Spread in Salaries
high_spread = clean_df.sort_values('Spread', ascending=False)
high_spread[['Undergraduate Major', 'Spread']].head()
```

Fun notice from above:
Notice how 3 of the top 5 are present in both. This means that there are some very high earning Economics degree holders, 
but also some who are not earning as much.

## Grouping and Pivoting Data with Pandas

Often times you will want to sum rows that belong to a particular category.
For example, which category of degrees has the highest average salary? Is it STEM, Business or HASS (Humanities, Arts, and Social Science)? 

To answer this question we can to use the .groupby() method.

```python
# For example, We have three categories in the 'Group' column: STEM, HASS and Business. 
# Let's count how many majors we have in each category:
clean_df.groupby('Group').count()
```

```python
# Formattinng better
pd.options.display.float_format = '{:,.2f}'.format 

# Now using the .mean() method to find the average salary by group? 
clean_df.groupby('Group').mean(numeric_only=True)
```
